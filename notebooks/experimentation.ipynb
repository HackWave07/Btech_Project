```python
import sys, os
# Add project root to path
ROOT = os.path.abspath('..')
if ROOT not in sys.path:
    sys.path.insert(0, ROOT)

import torch
import matplotlib.pyplot as plt
from src.config.config import Config
print('Device:', Config.DEVICE)
print('Classes:', len(Config.ALLOWED_FOLDERS))
```

```python
# ── Dataset exploration ────────────────────────────────────────────────
import pandas as pd
csv = os.path.join(ROOT, 'dataset.csv')
if os.path.exists(csv):
    df = pd.read_csv(csv)
    display(df.head())
    df['class'].value_counts().plot(kind='bar', figsize=(12,4), title='Class Distribution')
    plt.tight_layout()
    plt.show()
else:
    print('Run: python main.py --mode prepare  to generate dataset.csv')
```

```python
# ── Load and inspect trained classifier ───────────────────────────────
from src.models.classifier import ResNetClassifier

ckpt = os.path.join(ROOT, 'models', 'checkpoints', 'best_classifier.pth')
if os.path.exists(ckpt):
    model = ResNetClassifier(len(Config.ALLOWED_FOLDERS), len(Config.SEVERITY_LEVELS))
    model.load_state_dict(torch.load(ckpt, map_location='cpu'))
    model.eval()
    print(model)
else:
    print('Checkpoint not found. Train the model first.')
```
